```python
import os
import zipfile
import urllib.request

# Create folder
os.makedirs("coco", exist_ok=True)

# COCO annotations URL
url = "http://images.cocodataset.org/annotations/annotations_trainval2017.zip"

# Download only if not already present
if not os.path.exists("coco/annotations_trainval2017.zip"):
    print("Downloading annotations...")
    urllib.request.urlretrieve(
        url,
        "coco/annotations_trainval2017.zip"
    )

# Extract ZIP file
print("Extracting files...")

with zipfile.ZipFile(
    "coco/annotations_trainval2017.zip",
    "r"
) as zip_ref:
    zip_ref.extractall("coco")

print("✅ COCO annotations downloaded and extracted successfully!")
```

```python
import os
import zipfile
import urllib.request

# COCO validation images URL
url = "http://images.cocodataset.org/zips/val2017.zip"

# Download only if not already present
if not os.path.exists("coco/val2017.zip"):
    print("Downloading validation images...")
    urllib.request.urlretrieve(
        url,
        "coco/val2017.zip"
    )

# Extract ZIP file
print("Extracting images...")

with zipfile.ZipFile(
    "coco/val2017.zip",
    "r"
) as zip_ref:
    zip_ref.extractall("coco")

print("✅ COCO validation images downloaded and extracted successfully!")
```

```python
from pycocotools.coco import COCO

# Path to caption annotations
ann_file = "coco/annotations/captions_val2017.json"

# Load COCO captions
coco = COCO(ann_file)

print("✅ COCO Dataset Loaded Successfully!")

# Total images
image_ids = coco.getImgIds()

print("Total Images:", len(image_ids))
```

```python
import numpy as np

captions = []
widths = []
heights = []

for img_id in image_ids:

    # Get captions
    ann_ids = coco.getAnnIds(imgIds=img_id)
    anns = coco.loadAnns(ann_ids)

    for ann in anns:
        captions.append(ann["caption"])

    # Get image information
    img_info = coco.loadImgs(img_id)[0]

    widths.append(img_info["width"])
    heights.append(img_info["height"])


# Caption lengths (number of words)
caption_lengths = [
    len(caption.split())
    for caption in captions
]

print("=" * 50)

print("Total Images:", len(image_ids))
print("Total Captions:", len(captions))

print("\nCaption Statistics:")
print("Average Length:", round(np.mean(caption_lengths), 2), "words")
print("Minimum Length:", min(caption_lengths), "words")
print("Maximum Length:", max(caption_lengths), "words")

print("\nImage Resolution Statistics:")
print("Average Width:", round(np.mean(widths), 2))
print("Average Height:", round(np.mean(heights), 2))

print("=" * 50)
```

```python
from pycocotools.coco import COCO

instance_file = "coco/annotations/instances_val2017.json"

coco_instance = COCO(instance_file)

categories = coco_instance.loadCats(
    coco_instance.getCatIds()
)

class_names = [
    cat["name"]
    for cat in categories
]

print("Total Classes:", len(class_names))

print("\nCOCO Categories:\n")

for i, name in enumerate(class_names, 1):
    print(f"{i}. {name}")
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import random

# Select 5 random images
sample_ids = random.sample(image_ids, 5)

for img_id in sample_ids:

    img_info = coco.loadImgs(img_id)[0]

    image_path = "coco/val2017/" + img_info["file_name"]

    image = Image.open(image_path)

    # Load captions
    ann_ids = coco.getAnnIds(imgIds=img_id)
    anns = coco.loadAnns(ann_ids)

    captions = [
        ann["caption"]
        for ann in anns
    ]

    plt.figure(figsize=(8, 6))

    plt.imshow(image)
    plt.axis("off")

    # Show first 3 captions
    plt.title(
        "\n".join(captions[:3]),
        fontsize=10
    )

    plt.show()
```
